```python
####Task 1 — File I/O

print("Task 1 - File Read & Write\n")

file_name = "python_notes.txt"

#### Part A — Write to File

notes = [
    "Topic 1: Variables store data. Python is dynamically typed.",
    "Topic 2: Lists are ordered and mutable.",
    "Topic 3: Dictionaries store key-value pairs.",
    "Topic 4: Loops automate repetitive tasks.",
    "Topic 5: Exception handling prevents crashes."
]

# Writing to file (overwrite mode)
with open(file_name, "w", encoding="utf-8") as file:
    for line in notes:
        file.write(line + "\n")

print("File written successfully.")

# Append additional lines
extra_notes = [
    "Topic 6: Functions help reuse code.",
    "Topic 7: File handling allows data persistence."
]

with open(file_name, "a", encoding="utf-8") as file:
    for line in extra_notes:
        file.write(line + "\n")

print("Lines appended successfully.")


#### Part B — Read File

print("\nReading file:\n")

with open(file_name, "r", encoding="utf-8") as file:
    lines = file.readlines()

# Print numbered lines
for i, line in enumerate(lines, start=1):
    print(f"{i}. {line.strip()}")

# Count total lines
print(f"\nTotal number of lines: {len(lines)}")

# Keyword Search

keyword = input("\nEnter keyword to search: ").lower()

found = False

for line in lines:
    if keyword in line.lower():
        print(line.strip())
        found = True

if not found:
    print("No matching lines found.")
```

```python
import requests

print("Task 2 - API Integration\n")

BASE_URL = "https://dummyjson.com/products"

# Step 1 — Fetch Products

response = requests.get(f"{BASE_URL}?limit=20")

data = response.json()
products = data["products"]

print("Product List:\n")
print(f"{'ID':<5} {'Title':<30} {'Category':<15} {'Price':<10} {'Rating':<6}")
print("-" * 75)

for p in products:
    print(f"{p['id']:<5} {p['title']:<30} {p['category']:<15} ${p['price']:<10} {p['rating']:<6}")

# Step 2 — Filter & Sort

# Filter products with rating >= 4.5
filtered = [p for p in products if p["rating"] >= 4.5]

# Sort by price (descending)
filtered_sorted = sorted(filtered, key=lambda x: x["price"], reverse=True)

print("\nHigh Rated Products (rating >= 4.5):\n")

for p in filtered_sorted:
    print(f"{p['title']} - ${p['price']} (Rating: {p['rating']})")

# Step 3 — Search by Category

laptop_response = requests.get(f"{BASE_URL}/category/laptops")
laptops = laptop_response.json()["products"]

print("\nLaptops Category:\n")

for laptop in laptops:
    print(f"{laptop['title']} - ${laptop['price']}")

# Step 4 — POST Request

new_product = {
    "title": "My Custom Product",
    "price": 999,
    "category": "electronics",
    "description": "A product I created via API"
}

post_response = requests.post(f"{BASE_URL}/add", json=new_product)

print("\nPOST Response:\n")
print(post_response.json())
```

```python
import requests

print("Task 3 - Exception Handling\n")

#### Part A — Guarded Calculator
print("Part A — Guarded Calculator\n")
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Cannot divide by zero"
    except TypeError:
        return "Error: Invalid input types"

# Testing
print("Calculator Tests:")
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide("ten", 2))


#### Part B — Guarded File Reader
print("\nPart B — Guarded File Reader")
def read_file_safe(filename):
    try:
        with open(filename, "r", encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    finally:
        print("File operation attempt complete.")

# Testing
print("\nFile Read Tests:")
content = read_file_safe("python_notes.txt")
if content:
    print(content)

read_file_safe("ghost_file.txt")



#### Part C — Robust API Calls
print("\nPart C — Robust API Calls")

BASE_URL = "https://dummyjson.com/products"

# Error Logging Function

def log_error(message):
    # writing error messages into a file
    with open("error_log.txt", "a", encoding="utf-8") as file:
        file.write(message + "\n")

# Safe GET Request

def safe_get(url):
    try:
        response = requests.get(url, timeout=5)
        return response.json()

    except requests.exceptions.ConnectionError:
        msg = "Connection failed. Please check your internet."
        print(msg)
        log_error(msg)

    except requests.exceptions.Timeout:
        msg = "Request timed out. Try again later."
        print(msg)
        log_error(msg)

    except Exception as e:
        msg = f"Unexpected error: {str(e)}"
        print(msg)
        log_error(msg)


# Safe POST Request

def safe_post(url, data):
    try:
        response = requests.post(url, json=data, timeout=5)
        return response.json()

    except requests.exceptions.ConnectionError:
        msg = "Connection failed during POST request."
        print(msg)
        log_error(msg)

    except requests.exceptions.Timeout:
        msg = "POST request timed out."
        print(msg)
        log_error(msg)

    except Exception as e:
        msg = f"Unexpected POST error: {str(e)}"
        print(msg)
        log_error(msg)

# Testing API Calls

print("\nTesting API Calls:\n")

# 1. Normal GET
data = safe_get(f"{BASE_URL}?limit=5")
print(data)

# 2. Normal POST
new_product = {
    "title": "My Custom Product",
    "price": 999
}
print(safe_post(f"{BASE_URL}/add", new_product))


#### Part D — Input Validation Loop

print("\nPart D — Input Validation Loop")
print("\nProduct Lookup:")

while True:
    user_input = input("Enter a product ID (1–100) or 'quit': ")

    if user_input.lower() == "quit":
        print("Exiting...")
        break

    # Validate integer input
    if not user_input.isdigit():
        print("Invalid input. Please enter a number.")
        continue

    product_id = int(user_input)

    if product_id < 1 or product_id > 100:
        print("Please enter a number between 1 and 100.")
        continue

    # API Call
    try:
        response = requests.get(f"{BASE_URL}/{product_id}", timeout=5)

        if response.status_code == 200:
            data = response.json()
            print(f"{data['title']} — ${data['price']}")
        elif response.status_code == 404:
            print("Product not found.")

    except requests.exceptions.ConnectionError:
        print("Connection failed. Please check your internet.")
    except requests.exceptions.Timeout:
        print("Request timed out. Try again later.")
    except Exception as e:
        print("Error:", e)
```

```python
from datetime import datetime
import requests

LOG_FILE = "error_log.txt"

# Logger Function

def write_log(function_name, error_type, message):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    log_entry = f"[{timestamp}] ERROR in {function_name}: {error_type} — {message}"
    
    with open(LOG_FILE, "a", encoding="utf-8") as file:
        file.write(log_entry + "\n")


# 1. Trigger ConnectionError

def fetch_products():
    try:
        # invalid URL to force connection error
        response = requests.get("https://this-host-does-not-exist-xyz.com/api", timeout=5)
        return response.json()
    
    except requests.exceptions.ConnectionError as e:
        print("Connection error occurred")
        write_log("fetch_products", "ConnectionError", str(e))


# 2. Trigger HTTP 404 Error

def lookup_product(product_id):
    try:
        response = requests.get(f"https://dummyjson.com/products/{product_id}", timeout=5)
        
        # HTTP error handling (not exception)
        if response.status_code != 200:
            msg = f"{response.status_code} Not Found for product ID {product_id}"
            print("Product not found")
            write_log("lookup_product", "HTTPError", msg)
        else:
            data = response.json()
            print(data["title"], data["price"])
    
    except Exception as e:
        write_log("lookup_product", "UnexpectedError", str(e))


# Run Tests

fetch_products()        # will create ConnectionError log
lookup_product(999)     # will create HTTPError log

# 3. Print Log File Content

print("\n--- Error Log Contents ---\n")

with open(LOG_FILE, "r", encoding="utf-8") as file:
    print(file.read())
```
